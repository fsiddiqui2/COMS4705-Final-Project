```python
import json

def load_jsonl_native(filepath):
    data = []
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            data.append(json.loads(line.strip()))
    return data

loaded_data = load_jsonl_native('hotpot_closedbook_reasoning_10000.jsonl')
print(loaded_data)
```
